## Python 对象、基本类型和变量

Python 中的一切都是**对象**，每个对象都有一个**类型**。一些基本类型包括：

- **`int`**（整数；没有小数点的整数）
  - `10`
  - `-3`
- **`float`**（浮点数；有小数点的数字）
  - `7.41`
  - `-0.006`
- **`str`**（字符串；用单引号、双引号或三引号包围的字符序列）
  - `'这是一个使用单引号的字符串'`
  - `"这是一个使用双引号的字符串"`
  - `'''这是一个使用单引号的三引号字符串'''`
  - `"""这是一个使用双引号的三引号字符串"""`
- **`bool`**（布尔值；真或假的二进制值）
  - `True`
  - `False`
- **`NoneType`**（表示值缺失的特殊类型）
  - `None`

在 Python 中，**变量**是您在代码中指定的名称，它映射到特定的**对象**、对象**实例**或值。

通过定义变量，我们可以用对我们有意义的名称来引用事物。变量名只能包含字母、下划线（`_`）或数字（不能有空格、破折号或其他字符）。变量名必须以字母或下划线开头。

## 基本操作符

在 Python 中，有不同类型的**操作符**（特殊符号）对不同的值进行操作。一些基本操作符包括：

- 算术操作符
  - **`+`**（加法）
  - **`-`**（减法）
  - **`*`**（乘法）
  - **`/`**（除法）
  - **`**`**（幂运算）
- 赋值操作符
  - **`=`**（赋值）
  - **`+=`**（相加并重新赋值；递增）
  - **`-=`**（相减并重新赋值；递减）
  - **`*=`**（相乘并重新赋值）
- 比较操作符（返回 `True` 或 `False`）
  - **`==`**（等于）
  - **`!=`**（不等于）
  - **`<`**（小于）
  - **`<=`**（小于等于）
  - **`>`**（大于）
  - **`>=`**（大于等于）

当在单个表达式中使用多个操作符时，**操作符优先级**决定表达式的哪些部分按什么顺序求值。优先级较高的操作符先求值（就像数学中的 PEMDAS）。相同优先级的操作符从左到右求值。

- `()` 括号，用于分组
- `**` 幂运算
- `*`, `/` 乘法和除法
- `+`, `-` 加法和减法
- `==`, `!=`, `<`, `<=`, `>`, `>=` 比较

> 参考：https://docs.python.org/3/reference/expressions.html#operator-precedence

```python
# 将一些数字赋值给不同的变量
num1 = 10      # 整数：10
num2 = -3      # 整数：-3
num3 = 7.41    # 浮点数：7.41
num4 = -.6     # 浮点数：-0.6（可以省略小数点前的0）
num5 = 7       # 整数：7
num6 = 3       # 整数：3
num7 = 11.11   # 浮点数：11.11
```

```python
# 加法运算：10 + (-3) = 7
num1 + num2
```

```python
# 减法运算：-3 - 7.41 = -10.41
num2 - num3
```

```python
# 乘法运算：7.41 * (-0.6) = -4.446
num3 * num4
```

```python
# 除法运算：-0.6 / 7 = -0.08571428571428572
num4 / num5
```

```python
# 幂运算：7 的 3 次方 = 343
num5 ** num6
```

```python
# 递增现有变量：num7 = num7 + 4
num7 += 4  # 等价于 num7 = num7 + 4
num7       # 显示更新后的值
```

```python
# 递减现有变量：num6 = num6 - 2
num6 -= 2  # 等价于 num6 = num6 - 2
num6       # 显示更新后的值
```

```python
# 相乘并重新赋值：num3 = num3 * 5
num3 *= 5  # 等价于 num3 = num3 * 5
num3       # 显示更新后的值
```

```python
# 将表达式的值赋给变量（注意操作符优先级）
num8 = num1 + num2 * num3  # 先算乘法，再算加法
num8                       # 显示计算结果
```

```python
# 这两个表达式是否相等？
num1 + num2 == num5  # (10 + (-3)) == 7，结果为 True
```

```python
# 这两个表达式是否不相等？
num3 != num4  # 37.05 != -0.6，结果为 True
```

```python
# 第一个表达式是否小于第二个表达式？
num5 < num6  # 7 < 1，结果为 False
```

```python
# 这个表达式是否为真？（链式比较）
5 > 3 > 1  # 等价于 (5 > 3) and (3 > 1)，结果为 True
```

```python
# 这个表达式是否为真？（复杂链式比较）
5 > 3 < 4 == 3 + 1  # 等价于 (5 > 3) and (3 < 4) and (4 == 4)，结果为 True
```

```python
# 将一些字符串赋值给不同的变量
simple_string1 = 'an example'  # 使用单引号的字符串
simple_string2 = "oranges "    # 使用双引号的字符串（注意末尾有空格）
```

```python
# 字符串加法（连接）
simple_string1 + ' of using the + operator'  # 字符串连接
```

```python
# 注意：原字符串没有被修改（字符串是不可变的）
simple_string1  # 仍然是 'an example'
```

```python
# 字符串乘法（重复）
simple_string2 * 4  # 将字符串重复4次
```

```python
# 这个字符串也没有被修改
simple_string2  # 仍然是 "oranges "
```

```python
# 这两个表达式是否相等？
simple_string1 == simple_string2  # 'an example' == 'oranges '，结果为 False
```

```python
# 这两个表达式是否相等？
simple_string1 == 'an example'  # 'an example' == 'an example'，结果为 True
```

```python
# 字符串相加并重新赋值
simple_string1 += ' that re-assigned the original string'  # 修改原变量
simple_string1  # 显示更新后的值
```

```python
# 字符串相乘并重新赋值
simple_string2 *= 3  # 将字符串重复3次并重新赋值
simple_string2       # 显示更新后的值
```

## 基本容器

> 注意：**可变**对象在创建后可以被修改，**不可变**对象则不能。

容器是用于将其他对象组合在一起的对象。基本容器类型包括：

- **`str`**（字符串：不可变；通过整数索引；项目按添加顺序存储）
- **`list`**（列表：可变；通过整数索引；项目按添加顺序存储）
  - `[3, 5, 6, 3, 'dog', 'cat', False]`
- **`tuple`**（元组：不可变；通过整数索引；项目按添加顺序存储）
  - `(3, 5, 6, 3, 'dog', 'cat', False)`
- **`set`**（集合：可变；无索引；项目不按添加顺序存储；只能包含不可变对象；不包含重复对象）
  - `{3, 5, 6, 3, 'dog', 'cat', False}`
- **`dict`**（字典：可变；键值对通过不可变键索引；项目不按添加顺序存储）
  - `{'name': 'Jane', 'age': 23, 'fav_foods': ['pizza', 'fruit', 'fish']}`

定义列表、元组或集合时，使用逗号（,）分隔各个项目。定义字典时，使用冒号（:）分隔键和值，使用逗号（,）分隔键值对。

字符串、列表和元组都是**序列类型**，可以使用 `+`、`*`、`+=` 和 `*=` 操作符。

```python
# 将一些容器赋值给不同的变量
list1 = [3, 5, 6, 3, 'dog', 'cat', False]  # 列表：可变，有序，可重复
tuple1 = (3, 5, 6, 3, 'dog', 'cat', False)  # 元组：不可变，有序，可重复
set1 = {3, 5, 6, 3, 'dog', 'cat', False}  # 集合：可变，无序，不重复
dict1 = {'name': 'Jane', 'age': 23, 'fav_foods': ['pizza', 'fruit', 'fish']}  # 字典：键值对
```

```python
# 列表对象中的项目按添加顺序存储
list1  # 显示列表内容
```

```python
# 元组对象中的项目按添加顺序存储
tuple1  # 显示元组内容
```

```python
# 集合对象中的项目不按添加顺序存储
# 另外，注意值3在这个集合对象中只出现一次（集合不包含重复项）
set1  # 显示集合内容
```

```python
# 字典对象中的项目不按添加顺序存储
dict1  # 显示字典内容
```

```python
# 列表相加并重新赋值
list1 += [5, 'grapes']  # 在列表末尾添加新元素
list1  # 显示更新后的列表
```

```python
# 元组相加并重新赋值
tuple1 += (5, 'grapes')  # 创建新的元组（元组不可变）
tuple1  # 显示更新后的元组
```

```python
# 列表乘法（重复）
[1, 2, 3, 4] * 2  # 将列表重复2次
```

```python
# 元组乘法（重复）
(1, 2, 3, 4) * 3  # 将元组重复3次
```

## 访问容器中的数据

对于字符串、列表、元组和字典，我们可以使用**下标表示法**（方括号）来访问索引处的数据。

- 字符串、列表和元组通过整数索引，**从0开始**作为第一项
  - 这些序列类型还支持访问项目范围，称为**切片**
  - 使用**负索引**从序列的后面开始
- 字典通过其键进行索引

> 注意：集合没有索引，因此我们不能使用下标表示法来访问数据元素。

```python
# 访问序列中的第一个项目
list1[0]  # 获取索引0处的元素（第一个元素）
```

```python
# 访问序列中的最后一个项目
tuple1[-1]  # 使用负索引获取最后一个元素
```

```python
# 访问序列中一定范围的项目（切片）
simple_string1[3:8]  # 获取从索引3到7的字符（不包含索引8）
```

```python
# 访问序列中一定范围的项目（切片）
tuple1[:-3]  # 获取除最后3个元素外的所有元素
```

```python
# 访问序列中一定范围的项目（切片）
list1[4:]  # 获取从索引4开始到末尾的所有元素
```

```python
# 访问字典中的项目
dict1['name']  # 通过键'name'获取对应的值
```

```python
# 访问字典中序列的元素
dict1['fav_foods'][2]  # 先获取'fav_foods'键的值（列表），再获取索引2处的元素
```

## Python 内置函数和可调用对象

**函数**是您可以"调用"来**执行操作**或计算并**返回另一个对象**的 Python 对象。您通过在函数名右侧放置括号来调用函数。某些函数允许您在括号内传递**参数**（用逗号分隔多个参数）。在函数内部，这些参数被视为变量。

Python 有几个有用的内置函数来帮助您处理不同的对象和/或您的环境。以下是其中的一小部分：

- **`type(obj)`** 确定对象的类型
- **`len(container)`** 确定容器中有多少项目
- **`callable(obj)`** 确定对象是否可调用
- **`sorted(container)`** 从容器返回一个新列表，项目已排序
- **`sum(container)`** 计算数字容器的总和
- **`min(container)`** 确定容器中的最小项目
- **`max(container)`** 确定容器中的最大项目
- **`abs(number)`** 确定数字的绝对值
- **`repr(obj)`** 返回对象的字符串表示

> 完整的内置函数列表：https://docs.python.org/3/library/functions.html

还有定义您自己的函数和可调用对象的不同方法，我们稍后将探讨。

```python
# 使用 type() 函数确定对象的类型
type(simple_string1)  # 返回字符串对象的类型
```

```python
# 使用 len() 函数确定容器中有多少项目
len(dict1)  # 返回字典中键值对的数量
```

```python
# 使用 len() 函数确定容器中有多少项目
len(simple_string2)  # 返回字符串中字符的数量
```

```python
# 使用 callable() 函数确定对象是否可调用
callable(len)  # 检查 len 函数是否可调用
```

```python
# 使用 callable() 函数确定对象是否可调用
callable(dict1)  # 检查字典对象是否可调用
```

```python
# 使用 sorted() 函数从容器返回一个新列表，项目已排序
sorted([10, 1, 3.6, 7, 5, 2, -3])  # 对数字列表进行排序
```

```python
# 使用 sorted() 函数从容器返回一个新列表，项目已排序
# - 注意大写字符串排在前面
sorted(['dogs', 'cats', 'zebras', 'Chicago', 'California', 'ants', 'mice'])  # 对字符串列表进行排序
```

```python
# 使用 sum() 函数计算数字容器的总和
sum([10, 1, 3.6, 7, 5, 2, -3])  # 计算列表中所有数字的总和
```

```python
# 使用 min() 函数确定容器中的最小项目
min([10, 1, 3.6, 7, 5, 2, -3])  # 找到列表中的最小值
```

```python
# 使用 max() 函数确定容器中的最大项目
max('gibberish')  # 找到字符串中的最大字符（按字母顺序）
```

```python
# 使用 abs() 函数确定数字的绝对值
abs(10)  # 正数的绝对值是它本身
```

```python
# 使用 abs() 函数确定数字的绝对值
abs(-12)  # 负数的绝对值是它的相反数
```

## Python 对象属性（方法和属性）

Python 中不同类型的对象有不同的**属性**，可以通过名称引用（类似于变量）。要访问对象的属性，请在对象后使用点（`.`），然后指定属性（即 `obj.attribute`）

当对象的属性是可调用的时，该属性称为**方法**。它与函数相同，只是这个函数绑定到特定对象。

当对象的属性不可调用时，该属性称为**属性**。它只是关于对象的数据，本身是另一个对象。

内置的 `dir()` 函数可用于返回对象属性的列表。

## 字符串对象的一些方法

- **`.capitalize()`** 返回字符串的大写版本（只有第一个字符大写）
- **`.upper()`** 返回字符串的大写版本（所有字符大写）
- **`.lower()`** 返回字符串的小写版本（所有字符小写）
- **`.count(substring)`** 返回子字符串在字符串中出现的次数
- **`.startswith(substring)`** 确定字符串是否以子字符串开头
- **`.endswith(substring)`** 确定字符串是否以子字符串结尾
- **`.replace(old, new)`** 返回字符串的副本，其中"old"的出现被"new"替换

```python
# 将字符串赋值给变量
a_string = 'tHis is a sTriNg'  # 包含大小写混合的字符串
```

```python
# 返回字符串的大写版本（只有第一个字符大写）
a_string.capitalize()  # 将第一个字符大写，其余小写
```

```python
# 返回字符串的大写版本（所有字符大写）
a_string.upper()  # 将所有字符转换为大写
```

```python
# 返回字符串的小写版本（所有字符小写）
a_string.lower()  # 将所有字符转换为小写
```

```python
# 注意：调用的方法实际上没有修改原字符串（字符串是不可变的）
a_string  # 原字符串保持不变
```

```python
# 计算子字符串在字符串中出现的次数
a_string.count('i')  # 计算字符'i'在字符串中出现的次数
```

```python
# 计算子字符串在字符串中某个位置之后出现的次数
a_string.count('i', 7)  # 从索引7开始计算字符'i'的出现次数
```

```python
# 计算子字符串在字符串中出现的次数
a_string.count('is')  # 计算子字符串'is'在字符串中出现的次数
```

```python
# 字符串是否以'this'开头？
a_string.startswith('this')  # 检查字符串是否以'this'开头（区分大小写）
```

```python
# 小写字符串是否以'this'开头？
a_string.lower().startswith('this')  # 先转换为小写，再检查是否以'this'开头
```

```python
# 字符串是否以'Ng'结尾？
a_string.endswith('Ng')  # 检查字符串是否以'Ng'结尾（区分大小写）
```

```python
# 返回字符串的版本，其中子字符串被其他内容替换
a_string.replace('is', 'XYZ')  # 将所有'is'替换为'XYZ'
```

```python
# 返回字符串的版本，其中子字符串被其他内容替换
a_string.replace('i', '!')  # 将所有'i'替换为'!'
```

```python
# 返回字符串的版本，其中前2个出现的子字符串被其他内容替换
a_string.replace('i', '!', 2)  # 只替换前2个'i'为'!'
```

```python
# 2. 文本处理和提示词工程
def create_prompt(user_input, system_message="你是一个有用的AI助手。"):
    """
    创建结构化的提示词
    这是大模型应用中的核心技能
    """
    prompt = f"""
系统消息: {system_message}

用户输入: {user_input}

请根据以上信息提供帮助。
"""
    return prompt.strip()

# 使用示例
user_question = "请解释什么是机器学习？"
prompt = create_prompt(user_question)
print("生成的提示词:")
print(prompt)
```

```python
# 3. 列表推导式和数据处理（大模型数据处理常用）
# 处理多个对话历史
conversation_history = [
    {"role": "user", "content": "你好"},
    {"role": "assistant", "content": "你好！有什么可以帮助您的吗？"},
    {"role": "user", "content": "请介绍一下Python"},
    {"role": "assistant", "content": "Python是一种高级编程语言..."}
]

# 使用列表推导式提取所有用户消息
user_messages = [msg["content"] for msg in conversation_history if msg["role"] == "user"]
print("用户消息:", user_messages)

# 使用列表推导式计算每条消息的长度
message_lengths = [len(msg["content"]) for msg in conversation_history]
print("消息长度:", message_lengths)

# 使用字典推导式创建角色到消息的映射
role_messages = {msg["role"]: [m["content"] for m in conversation_history if m["role"] == msg["role"]] 
                 for msg in conversation_history}
print("按角色分组的消息:", role_messages)
```

```python
# 4. 错误处理和异常处理（大模型API调用必备）
def safe_api_call(api_function, *args, **kwargs):
    """
    安全地调用API函数，包含错误处理
    这是大模型应用中的最佳实践
    """
    try:
        result = api_function(*args, **kwargs)
        return {"success": True, "data": result, "error": None}
    except ConnectionError as e:
        return {"success": False, "data": None, "error": f"连接错误: {e}"}
    except ValueError as e:
        return {"success": False, "data": None, "error": f"值错误: {e}"}
    except Exception as e:
        return {"success": False, "data": None, "error": f"未知错误: {e}"}

# 模拟API调用函数
def mock_llm_api_call(prompt):
    """模拟LLM API调用"""
    if not prompt:
        raise ValueError("提示词不能为空")
    if len(prompt) > 1000:
        raise ConnectionError("提示词过长，连接超时")
    return f"AI回复: {prompt[:50]}..."

# 测试安全API调用
test_prompts = ["你好", "", "很长的提示词" * 100]

for prompt in test_prompts:
    result = safe_api_call(mock_llm_api_call, prompt)
    print(f"提示词: '{prompt[:20]}...'")
    print(f"结果: {result}")
    print("-" * 40)
```

```python
# 5. 类和对象（大模型应用架构设计）
class LLMClient:
    """
    大语言模型客户端类
    展示面向对象编程在大模型应用中的应用
    """
    def __init__(self, model_name="gpt-3.5-turbo", api_key=None):
        self.model_name = model_name
        self.api_key = api_key
        self.conversation_history = []
    
    def add_message(self, role, content):
        """添加消息到对话历史"""
        self.conversation_history.append({
            "role": role,
            "content": content
        })
    
    def get_conversation_context(self, max_messages=10):
        """获取对话上下文（最近的消息）"""
        return self.conversation_history[-max_messages:]
    
    def clear_history(self):
        """清空对话历史"""
        self.conversation_history = []
    
    def __str__(self):
        return f"LLMClient(model={self.model_name}, messages={len(self.conversation_history)})"

# 使用示例
client = LLMClient("gpt-4", "your-api-key")
client.add_message("user", "你好")
client.add_message("assistant", "你好！有什么可以帮助您的吗？")
client.add_message("user", "请介绍一下Python")

print("客户端信息:", client)
print("对话上下文:", client.get_conversation_context())
```

## Positional arguments and keyword arguments to callables

You can call a function/method in a number of different ways:

- `func()`: Call `func` with no arguments
- `func(arg)`: Call `func` with one positional argument
- `func(arg1, arg2)`: Call `func` with two positional arguments
- `func(arg1, arg2, ..., argn)`: Call `func` with many positional arguments
- `func(kwarg=value)`: Call `func` with one keyword argument 
- `func(kwarg1=value1, kwarg2=value2)`: Call `func` with two keyword arguments
- `func(kwarg1=value1, kwarg2=value2, ..., kwargn=valuen)`: Call `func` with many keyword arguments
- `func(arg1, arg2, kwarg1=value1, kwarg2=value2)`: Call `func` with positonal arguments and keyword arguments
- `obj.method()`: Same for `func`.. and every other `func` example

When using **positional arguments**, you must provide them in the order that the function defined them (the function's **signature**).

When using **keyword arguments**, you can provide the arguments you want, in any order you want, as long as you specify each argument's name.

When using positional and keyword arguments, positional arguments must come first.

## Python "for loops"

It is easy to **iterate** over a collection of items using a **for loop**. The strings, lists, tuples, sets, and dictionaries we defined are all **iterable** containers.

The for loop will go through the specified container, one item at a time, and provide a temporary variable for the current item. You can use this temporary variable like a normal variable.

## Python "if statements" and "while loops"

Conditional expressions can be used with these two **conditional statements**.

The **if statement** allows you to test a condition and perform some actions if the condition evaluates to `True`. You can also provide `elif` and/or `else` clauses to an if statement to take alternative actions if the condition evaluates to `False`.

The **while loop** will keep looping until its conditional expression evaluates to `False`.

> Note: It is possible to "loop forever" when using a while loop with a conditional expression that never evaluates to `False`.
>
> Note: Since the **for loop** will iterate over a container of items until there are no more, there is no need to specify a "stop looping" condition.
